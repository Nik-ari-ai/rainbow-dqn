## 4. Реализация и настройка — участник 2

- **Реализация:** [Kaixhin/Rainbow](https://github.com/Kaixhin/Rainbow), commit `1745b18`, лицензия MIT (копия в `src/LICENSE-Kaixhin-Rainbow.md`).
- **Все шесть компонентов есть:** Double DQN и distributional C51 (`src/agent.py`), Dueling и Noisy Nets (`src/model.py`), Prioritized Replay и multi-step n=3 (`src/memory.py`).
- **Что изменено:** `atari-py` заменён на `ale-py` (`src/env.py`, `src/main.py`), остальное без изменений.
- **Зависимости:** `requirements.txt`; PyTorch с CUDA 12.8 (нужен для RTX 5070), команды в `README.md`.
- **Среда:** `space_invaders` (ALE, ROM в пакете `ale-py`), wrappers внутри `src/env.py`: серый кадр 84×84, стек 4 кадра, frame skip 4 с max-pool, до 30 случайных no-op, потеря жизни = конец эпизода при обучении, клиппинг награды в [-1, 1].
- **Гиперпараметры:** по умолчанию из статьи Rainbow (lr 6.25e-5, batch 32, γ=0.99, n=3, 51 атом, V∈[-10, 10], target update 8k, replay 1M); seed 123; устройство cuda, если доступно.

## 5. Обучение или загрузка модели — участник 2

Обучение запускается командой из `README.md` (на компьютере с GPU). Ниже — загрузка весов и выбор действия.
Веса: `results/spaceinv/model.pth` (TODO: указать итоговое число шагов обучения и ссылку, когда обучение завершится).

```python
args = argparse.Namespace(
    game='space_invaders', seed=123, device=torch.device('cuda' if torch.cuda.is_available() else 'cpu'),
    max_episode_length=int(108e3), history_length=4, architecture='canonical', hidden_size=512, noisy_std=0.1,
    atoms=51, V_min=-10, V_max=10, batch_size=32, multi_step=3, discount=0.99, norm_clip=10,
    learning_rate=6.25e-5, adam_eps=1.5e-4, model='../results/spaceinv/model.pth')

env = Env(args); env.eval()
agent = Agent(args, env); agent.eval()      # загружает веса из args.model
state = env.reset()
action = agent.act(state)                   # greedy-действие для одного наблюдения
print('observation:', tuple(state.shape), '| action:', action, '/', env.action_space())
```
